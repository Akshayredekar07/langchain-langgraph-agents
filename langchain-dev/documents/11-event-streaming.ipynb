```python
import os

from dotenv import load_dotenv
from langchain_nebius import ChatNebius
from langchain_nvidia_ai_endpoints import ChatNVIDIA


load_dotenv()


# Validate that the required environment variables exist.
if not os.getenv("NEBIUS_API_KEY"):
    raise ValueError("NEBIUS_API_KEY is not set")

if not os.getenv("NVIDIA_API_KEY"):
    raise ValueError("NVIDIA_API_KEY is not set")


# Nebius
# ChatNebius automatically reads NEBIUS_API_KEY
nebius_model = ChatNebius(
    model="zai-org/GLM-5.2",
    temperature=0.3,
)


# NVIDIA
# ChatNVIDIA automatically reads NVIDIA_API_KEY
reasoning_model = ChatNVIDIA(
    # model="deepseek-ai/deepseek-v4-pro-0813",
    model="moonshotai/kimi-k3",
    temperature=0.3,
)
```
